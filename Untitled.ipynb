```python
from coffea.util import load
import matplotlib.pyplot as plt
import mplhep as hep
import hist
```

```python
output = load("hists_QCD_sf_test_Zb/hists_QCD_sf_test_Zb.coffea")
```

```python
plt.style.use(hep.style.ROOT)
```

```python
output
```

```python
ds = "DY_2J_amcnlo"

for label, key in (("Overall", "cutflow"), ("Zee channel", "cutflow_Zee"), ("Zmm channel", "cutflow_Zmm")):
    print(f"=== {label} ===")
    for cut, count in output[ds][key].items():
        print(f"{cut:20s} {count}")
    print()
```

```python
h = output[ds]["mu0_pt"]
h[{"syst": "nominal", "channel": "zmm"}].plot1d()
plt.xlabel("Leading muon $p_T$ [GeV]")
plt.show()
```

```python
h = output[ds]["mu0_pt"]
h_sel = h[{"syst": "nominal", "channel": "zmm"}]

h_sel[::hist.rebin(10)].plot1d()   # merge every 4 bins into 1
plt.xlabel("Leading muon $p_T$ [GeV]")
plt.show()
```
